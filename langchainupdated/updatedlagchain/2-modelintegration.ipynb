## Groq model Integration 


```python
from langchain.chat_models import init_chat_model

# Use colon (provider:model), not slash — e.g. groq:llama-3.3-70b-versatile
model = init_chat_model("groq:llama-3.3-70b-versatile")

response = model.invoke("what is the name of maharashtra capital?")
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="llama-3.3-70b-versatile")

response = model.invoke("what is the name of maharashtra capital?")
response.content
```

## Streaming 
##### In the case of streaming we can stream their output content while it is being generated 

```python
model.invoke("write 200 words praragraph on artifical intelligence")
```

```python
# end="" keeps chunks on the same line (paragraph); default print() adds newline after each token
for chunk in model.stream("write me a 200 words paragraph on artifical intelligence?"):
    print(chunk.text, end="", flush=True)
print()
```

```python
# end="" joins chunks into one line; end="|" was only for demo (adds | between every token)
for chunk in model.stream("write me a 200 words paragraph on artifical intelligence?"):
    print(chunk.text, end="", flush=True)
print()  
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)
print()  
```
